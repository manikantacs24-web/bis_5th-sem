```python

import numpy as np
import matplotlib.pyplot as plt

X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 1, 1, 0])

population_size = 10
generations = 50
mutation_rate = 0.1

population = np.random.randn(population_size, 2)

def fitness(weights):
    prediction = X @ weights
    error = np.mean((y - prediction) ** 2)
    return error

best_errors = []

for generation in range(generations):

    scores = np.array([fitness(w) for w in population])


    best_index = np.argmin(scores)
    best = population[best_index].copy()

    best_errors.append(scores[best_index])

    new_population = [best]

    while len(new_population) < population_size:

        parent = population[np.random.randint(population_size)]

        child = (parent + best) / 2

        if np.random.rand() < mutation_rate:
            child += np.random.randn(2) * 0.1

        new_population.append(child)

    population = np.array(new_population)

print("Best weights:", best)
print("Minimum error:", round(float(best_errors[-1]), 4))

plt.plot(best_errors)
plt.xlabel("Generation")
plt.ylabel("Error")
plt.title("Genetic Algorithm Optimization")
plt.show()
```
