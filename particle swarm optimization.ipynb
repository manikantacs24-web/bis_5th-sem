```python

import random
import math


def griewank(x):
    s = sum(xi**2 for xi in x) / 4000
    p = 1
    for i in range(len(x)):
        p *= math.cos(x[i] / math.sqrt(i + 1))
    return 1 + s - p


n = 10
dim = 2
iterations = 100
w = 0.5
c1 = 1.5
c2 = 1.5


pos = [[random.uniform(-10, 10) for _ in range(dim)]
       for _ in range(n)]
vel = [[random.uniform(-1, 1) for _ in range(dim)]
       for _ in range(n)]

pbest = [p[:] for p in pos]
gbest = min(pbest, key=griewank).copy()


for t in range(iterations):
    for i in range(n):
        for j in range(dim):
            r1 = random.random()
            r2 = random.random()

            vel[i][j] = (
                w * vel[i][j]
                + c1 * r1 * (pbest[i][j] - pos[i][j])
                + c2 * r2 * (gbest[j] - pos[i][j])
            )

            pos[i][j] += vel[i][j]

       
        if griewank(pos[i]) < griewank(pbest[i]):
            pbest[i] = pos[i][:]

 
    best = min(pbest, key=griewank)
    if griewank(best) < griewank(gbest):
        gbest = best.copy()


print("Best position:", gbest)
print("Minimum value:", griewank(gbest))
```
