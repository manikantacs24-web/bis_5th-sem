```python

# Parallel Cellular Algorithm for Robot Path Planning

# Grid size
ROWS = 7
COLS = 10

# Starting and goal positions
start = (0, 0)
goal = (6, 9)

# Obstacles
obstacles = [
    (1, 2), (2, 2), (3, 2), (4, 2),
    (1, 5), (2, 5), (3, 5),
    (4, 7), (5, 7)
]

INF = 9999

# Initialize cells
cost = [[INF for _ in range(COLS)]
        for _ in range(ROWS)]

cost[goal[0]][goal[1]] = 0

# Neighbor directions: Up, Down, Left, Right
directions = [
    (-1, 0), (1, 0),
    (0, -1), (0, 1)
]

# Update cells
for iteration in range(ROWS * COLS):

    new_cost = [row[:] for row in cost]
    changed = False

    for r in range(ROWS):
        for c in range(COLS):

            if (r, c) in obstacles:
                continue

            if (r, c) == goal:
                continue

            neighbors = []

            for dr, dc in directions:
                nr = r + dr
                nc = c + dc

                if (0 <= nr < ROWS and
                    0 <= nc < COLS and
                    (nr, nc) not in obstacles):

                    neighbors.append(cost[nr][nc])

            best = min(neighbors)

            if best < INF:
                new_cost[r][c] = best + 1

            if new_cost[r][c] != cost[r][c]:
                changed = True

    cost = new_cost

    if not changed:
        break

# Find robot path
path = [start]
current = start

while current != goal:

    r, c = current
    best_cost = cost[r][c]
    next_cell = None

    for dr, dc in directions:
        nr = r + dr
        nc = c + dc

        if (0 <= nr < ROWS and
            0 <= nc < COLS):

            if cost[nr][nc] < best_cost:
                best_cost = cost[nr][nc]
                next_cell = (nr, nc)

    if next_cell is None:
        print("No path found!")
        break

    current = next_cell
    path.append(current)

# Display result
if current == goal:

    print("Robot Path:", path)
    print("Total Steps:", len(path) - 1)



    for r in range(ROWS):
        for c in range(COLS):

            if (r, c) == start:
                print("S", end=" ")

            elif (r, c) == goal:
                print("G", end=" ")

            elif (r, c) in obstacles:
                print("#", end=" ")

            elif (r, c) in path:
                print("*", end=" ")

            else:
                print(".", end=" ")

        print()
```
